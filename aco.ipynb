```python
import numpy as np

# عدد المدن
NUM_CITIES = 5

# Distance Matrix
distance_matrix = np.array([
    [0, 2, 9, 10, 7],
    [2, 0, 6, 4, 3],
    [9, 6, 0, 8, 5],
    [10, 4, 8, 0, 6],
    [7, 3, 5, 6, 0]
])
```

```python
ALPHA = 1       # تأثير الفيرمون
BETA = 2        # تأثير المسافة (heuristic)
RHO = 0.5       # evaporation rate
Q = 100         # كمية الفيرمون
NUM_ANTS = 5
NUM_ITERATIONS = 10

# Initial pheromone
pheromone = np.ones((NUM_CITIES, NUM_CITIES)) # pheromone = 1 in begin 
```

```python
def choose_next_city(current, visited):
    probabilities = []

    for j in range(NUM_CITIES):
        if j not in visited:
            tau = pheromone[current][j] ** ALPHA
            eta = (1 / distance_matrix[current][j]) ** BETA
            probabilities.append((j, tau * eta))
    
    # Normalize
    total = sum(p for _, p in probabilities)
    probabilities = [(city, p / total) for city, p in probabilities]

    # Roulette Wheel
    r = np.random.rand()
    cumulative = 0

    for city, prob in probabilities:
        cumulative += prob
        if r <= cumulative:
            return city
```

```python
def construct_solution():
    path = [np.random.randint(NUM_CITIES)]

    while len(path) < NUM_CITIES:
        next_city = choose_next_city(path[-1], path)
        path.append(next_city)

    path.append(path[0])  # الرجوع للبداية
    return path
```

```python
def path_length(path):
    total = 0
    for i in range(len(path) - 1):
        total += distance_matrix[path[i]][path[i+1]]
    return total
```

```python
def update_pheromones(all_paths):
    global pheromone

    # Evaporation
    pheromone *= (1 - RHO)

    # Deposit
    for path, length in all_paths:
        for i in range(len(path) - 1):
            a, b = path[i], path[i+1]
            pheromone[a][b] += Q / length
            pheromone[b][a] += Q / length
```

```python
best_path = None
best_length = float('inf')

for iteration in range(NUM_ITERATIONS):
    all_paths = []

    for _ in range(NUM_ANTS):
        path = construct_solution()
        length = path_length(path)

        all_paths.append((path, length))

        if length < best_length:
            best_path = path
            best_length = length

    update_pheromones(all_paths)

    print(f"Iteration {iteration+1}: Best Path = {best_path}, Length = {best_length}")
```

```python
print("\nFinal Best Path:", best_path)
print("Final Best Length:", best_length)
```
